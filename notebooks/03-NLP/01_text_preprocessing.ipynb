# dataset_cleanup — text_preprocessing
pull both raw CSVs into pandas, layer the paidscraper columns, and output cleaned text table for next notebook.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / 'data').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.nlp.io import read_data_pd
from src.nlp.preprocess import build_preprocessed_frame
```

```python
base_path = PROJECT_ROOT / 'data/raw/bsr_visual_data.csv'
enriched_path = PROJECT_ROOT / 'data/raw/17k_products_amazon_data.csv'
base_df = read_data_pd(base_path)
enriched_df = read_data_pd(enriched_path)
base_df.shape, enriched_df.shape
```

```python
clean_df = build_preprocessed_frame(base_df, enriched_df)
clean_df.head()
```

```python
required_cols = ['asin','keyword','text_raw','text_clean','bsr_best','sd_number_bought_past_month','rating_pct_1','rating_pct_2','rating_pct_3','rating_pct_4','rating_pct_5','sentiment_count_positive','sentiment_count_mixed','sentiment_count_negative','flag_missing_keyword','flag_missing_brand','flag_missing_sd_title','image_token_count']
missing_cols = [col for col in required_cols if col not in clean_df.columns]
if missing_cols:
    raise ValueError(f'Missing expected columns: {missing_cols}')
clean_df[required_cols].head()
```

```python
clean_df['text_raw_len'] = clean_df['text_raw'].str.len().fillna(0).astype(int)
clean_df['text_clean_len'] = clean_df['text_clean'].str.len().fillna(0).astype(int)
length_summary = clean_df[['text_raw_len', 'text_clean_len']].describe().T
length_summary
```

```python
empty_ratio = (clean_df['text_clean_len'] == 0).mean()
print(f'empty text ratio: {empty_ratio:.3%}')
if empty_ratio > 0.05:
    raise ValueError('text_clean empty ratio above 5% – revisit preprocessing steps')
```

```python
missing_summary = clean_df[['flag_missing_keyword','flag_missing_brand','flag_missing_sd_title']].sum().rename('count')
missing_summary
```

```python
output_path = PROJECT_ROOT / 'data/interim/nlp_text_clean.parquet'
clean_df.to_parquet(output_path, index=False)
output_path
```
