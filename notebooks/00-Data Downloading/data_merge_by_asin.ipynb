```python
import pandas as pd
import os

# Load both datasets
df_40k = pd.read_csv('./../../data/final_datasets/40k_multisector_amazon_products.csv')
df_20k = pd.read_csv('./../../data/asin_products_20k.csv')

print(f"40k dataset shape: {df_40k.shape}")
print(f"20k dataset shape: {df_20k.shape}")

# Find common columns
common_columns = set(df_40k.columns) & set(df_20k.columns)
print(f"\nCommon columns: {len(common_columns)}")
print(f"Common columns: {sorted(common_columns)}")

# Keep only common columns
df_40k_common = df_40k[list(common_columns)].copy()
df_20k_common = df_20k[list(common_columns)].copy()

# Remove duplicates by ASIN, keeping first occurrence
df_40k_common = df_40k_common.drop_duplicates(subset=['asin'], keep='first')
df_20k_common = df_20k_common.drop_duplicates(subset=['asin'], keep='first')

print(f"\nAfter removing duplicates:")
print(f"40k unique ASINs: {len(df_40k_common)}")
print(f"20k unique ASINs: {len(df_20k_common)}")

# Check overlap between datasets
asin_40k_set = set(df_40k_common['asin'])
asin_20k_set = set(df_20k_common['asin'])
overlapping_asins = asin_40k_set & asin_20k_set
unique_to_40k = asin_40k_set - asin_20k_set
unique_to_20k = asin_20k_set - asin_40k_set

print(f"\nOverlap analysis:")
print(f"ASINs in both datasets: {len(overlapping_asins)}")
print(f"ASINs only in 40k dataset: {len(unique_to_40k)}")
print(f"ASINs only in 20k dataset: {len(unique_to_20k)}")

# Merge datasets - combine all unique ASINs
# When there are duplicates, keep the one from 40k dataset (first in concat)
merged_df = pd.concat([df_40k_common, df_20k_common], ignore_index=True)

# Remove duplicates again in case there are overlapping ASINs between datasets
# This keeps the first occurrence (from 40k dataset)
merged_df = merged_df.drop_duplicates(subset=['asin'], keep='first')

print(f"\nMerged dataset shape: {merged_df.shape}")
print(f"Total unique ASINs: {len(merged_df)}")
print(f"\nNote: {len(overlapping_asins)} ASINs were present in both datasets.")
print(f"      The version from the 40k dataset was kept for overlapping ASINs.")
print(f"      {len(unique_to_20k)} ASINs from 20k dataset were added (not in 40k).")

# Display first few rows
merged_df.head()
```

```python
merged_df.to_csv('./../../data/final_datasets/40k_products_no_images.csv')
```
